# Carbon_ETS — explore

Run `make run` from the repo root first to populate `data/`.

This notebook loads the panel and gives you a 5-minute tour. Copy any cell into your own notebook to start prototyping.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path

DATA = Path('..') / 'data'
panel = pd.read_parquet(DATA / 'panel_features.parquet')
print(panel.shape, panel.index.min().date(), '→', panel.index.max().date())
panel.head()
```

## 1. What's in the panel?

```python
for c in panel.columns:
    nn = panel[c].notna().sum()
    print(f'{c:35s}  non-null: {nn:5d}')
```

## 2. EUA proxy (KRBN) over time

```python
fig, ax = plt.subplots(figsize=(10,4))
panel['carbon_proxy_krbn'].plot(ax=ax, label='KRBN')
ax.set_title('KRBN (EUA proxy)  —  USD')
ax.grid(alpha=0.3)
```

## 3. Causal chain — visual sanity check

Industrial production (Eurostat IP YoY) vs EUA price. The chain says they should move together; the recession leg shows up clearly in 2020 (COVID) and 2022-23 (gas crisis demand destruction).

```python
fig, ax = plt.subplots(2,1, figsize=(10,6), sharex=True)
panel['carbon_proxy_krbn'].plot(ax=ax[0], color='C0'); ax[0].set_ylabel('KRBN'); ax[0].grid(alpha=0.3)
panel['ip_yoy'].plot(ax=ax[1], color='C3'); ax[1].axhline(0, color='k', lw=0.5)
ax[1].set_ylabel('IP YoY (EA19)'); ax[1].grid(alpha=0.3)
```

## 4. Backtest result — load the live equity curve

```python
bt = pd.read_parquet(DATA / 'backtest_trades.parquet')
fig, ax = plt.subplots(figsize=(10,4))
bt['equity'].plot(ax=ax, label='strategy')
(1+bt['ret']).cumprod().plot(ax=ax, color='grey', alpha=0.6, label='buy & hold')
ax.legend(); ax.set_title('Baseline equity curve'); ax.grid(alpha=0.3)
```
